```python
price_fields = ['open', 'high', 'low', 'close', 'vwap']
rel_fiedls = ["volume", "value", "openint"]
cover_cols=["volume", "value", "vwap"]
```

```python
instruments = 'rbb'
tick_size = 1
adjusted_method = None
begin_time = datetime.datetime(2026, 5, 6)
end_time = datetime.datetime(2026, 5, 13)

window =1 
weriod =2
ewm=True

method = 'ewm' if ewm else 'rolling'
```

```python
research_market, trader_market, metrics_data = fetch_market_data(
        instruments=instruments,
        begin_time=begin_time,
        end_time=end_time,
        tick_size=tick_size,
        adjusted_method=adjusted_method,
        price_fields=price_fields,
        rel_fiedls=rel_fiedls,
        cover_cols=cover_cols)
```

```python
research_res = market_data_format(research_market)
trader_res = market_data_format(trader_market)
```

```python
from lumina.impulse.fixed import *
```

```python
t_chg = safe_shift(trader_res['close'])
r_chg = safe_shift(research_res['close'])
factor_metrics(name='1', factor_research=r_chg.stack(), factor_trade=t_chg.stack())
```

```python
t_weight = roller_sum(t_chg * trader_res['openint'], weriod, 1, method)
r_weight = roller_sum(r_chg * research_res['openint'], weriod, 1, method)
factor_metrics(name='1', factor_research=r_weight.stack(), factor_trade=t_weight.stack())
```

```python
t_ont1 = roller_sum(trader_res['openint'], weriod, 1, method)
r_ont1 = roller_sum(research_res['openint'], weriod, 1, method)
factor_metrics(name='1', factor_research=r_ont1.stack(), factor_trade=t_ont1.stack())
```

```python
t_factor = -(t_weight / t_ont1)
r_factor = -(r_weight / r_ont1)
factor_metrics(name='1', factor_research=r_factor.stack(), factor_trade=t_factor.stack())
```

```python
t_alpha = roller_mean(t_factor, window, 1, method)
r_alpha = roller_mean(r_factor, window, 1, method)
factor_metrics(name='1', factor_research=r_alpha.stack(), factor_trade=t_alpha.stack())
```
