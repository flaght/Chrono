```python
import pandas  as pd
import numpy as np
import pdb, os, datetime, itertools, time, hashlib
import pyprojroot, sys
from pyprojroot.criterion import has_file
sys.path.insert(0, str(pyprojroot.find_root(has_file("pyproject.toml"))))

from dotenv import load_dotenv
load_dotenv()

from lib.flp001 import *
```

```python
method = 'ricso2'
task_id = '134001'
instruments = 'hcb'
period = 5
category='p'
```

```python
draft_data_recent = load_data3(method=method, task_id=task_id, instruments=instruments, 
          period=period, category=category,filename='cohort_pro.csv')
draft_data_recent['types'] = 'recent'
draft_data_recent.sort_values(by=['ann_sharpe'],ascending=False)[['formula','ic_mean','ann_sharpe','calmar','pl_ratio']]
```

```python
draft_data_train = load_data2(method=method, task_id=task_id, instruments=instruments, 
          period=period,filename='cohort_pro.csv')
draft_data_recent['types'] = 'train'
draft_data_train.sort_values(by=['ann_sharpe'],ascending=False)[['formula','ic_mean','ann_sharpe','calmar','pl_ratio']]
```

```python
draft_data = pd.concat([draft_data_train,draft_data_recent],axis=0).sort_values(by=['factor_id'])
```
