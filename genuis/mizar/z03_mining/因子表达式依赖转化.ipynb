```python
import os, math,random
import pandas as pd
```

```python
start_date_str = '2026-01-06'
end_date_str = '2026-06-01'
```

```python
base_postion = 1000
count = math.ceil(new_program.shape[0] / base_postion)
```

```python
current_date = pd.to_datetime(start_date_str)
end_date = pd.to_datetime(end_date_str)
```

```python
filenames = []

for _ in range(count):
    filenames.append(current_date.strftime('%Y%m%d'))
    step = random.randint(1, 3) 
    current_date += pd.Timedelta(days=step)
    if current_date > end_date:
        print(f"[警告] 生成的文件名日期已超过设定的 {end_date_str}！")
```

```python
os.makedirs(os.path.join("temp", "all_bak"), exist_ok=True)
```

```python
for i in range(0, count):
    # 使用 iloc 避免边界数据重复
    t1 = new_program.iloc[i * base_postion : (i+1) * base_postion]
    t2 = t1.reset_index(drop=True)
    
    # 取出刚刚生成的日期字符串作为文件名
    current_filename = filenames[i]
    save_path = os.path.join("temp", "all_bak", f"{current_filename}.feather")
    
    t2.to_feather(save_path)
```
