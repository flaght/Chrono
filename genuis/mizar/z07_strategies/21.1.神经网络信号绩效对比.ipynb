```python
from kdutils.tactix import Tactix
from kdutils.macro import base_path
from lib.flp001 import load_signal_performance,to_html1
```

```python
config_file = "rbb.yaml"
config_id = "nn_rbb_signal"
```

```python
args = argparse.Namespace(
    config_file=str(root / "config" / "research" / config_file),
    config_id=config_id,
)
```

```python
variant = Tactix().yaml(args)
```

```python
method = variant.method
instrument = variant.instruments
task_id = variant.task_id
period = variant.period
trial_id = variant.trial_id
types1 = "holdout" if  variant.form == "metrics" else "composite"
types2 = "test" if  variant.form == "metrics" else "val"
```
