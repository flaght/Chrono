```python
#### 对比公式
def rolling_compare(research_data, bench_data, name, window):
    common = (research_data[['trade_time', 'code', name
                             ]].merge(bench_data[['trade_time', 'code', name]],
                                      on=["trade_time", "code"],
                                      suffixes=("_research", "_bench")))
    common["rolling_corr"] = (
        common["{}_research".format(name)].rolling(window).corr(
            common["{}_bench".format(name)]))
    return common
```

### 成交量 价格对比

```python
volume_common = rolling_compare(
    research_data=research_data, 
    bench_data=bench_data, 
    name = 'volume', window=5)
volume_common.tail()
```

```python
volume_common[volume_common['rolling_corr'] < 0.9]
```

```python
openint_common = rolling_compare(
    research_data=research_data, 
    bench_data=bench_data, 
    name = 'openint', window=5)
openint_common.tail()
```
