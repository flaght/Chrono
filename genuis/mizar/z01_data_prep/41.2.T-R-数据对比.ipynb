#### 投研自行合成数据和交易自行合成数据对比

```python
import os
from dotenv import load_dotenv

load_dotenv()
import pandas as pd
```

```python
from config.contract import TRADING_TIME_MAPPING
from kdutils.data import fetch_local_market1, fetch_trader_market1
from lib.attr001.ftd001 import filter_trading_time
from lib.ret001 import create_chg, create_yields
```

```python
code = 'RB'
trading_sessions = TRADING_TIME_MAPPING[code]
begin_date = '2026-05-01'
end_data = '2026-07-01'
```

```python
keep_columns = ['trade_time','code','symbol', 'open','high','low','close','vwap','volume','openint']
```

```python
research_data = fetch_local_market1(base_path=os.environ['BAR_FUT_DIRS'],
                                       begin_date=begin_date,
                                       end_date=end_data,
                                       keep_symbol=True,
                                       codes=[code])
research_data = research_data[keep_columns]
research_data['trade_time'] = pd.to_datetime(research_data['trade_time'])
research_data = research_data.set_index(['trade_time','code'])
research_data = filter_trading_time(data=research_data, trading_sessions=trading_sessions).sort_values(by=['trade_time','code'])
```

```python
research_data.head()
```

```python
trader_data = fetch_trader_market1(begin_time=begin_date,
                                     end_time=end_data,
                                     code=code)
trader_data = trader_data[keep_columns]
trader_data['trade_time'] = pd.to_datetime(trader_data['trade_time'])
trader_data = trader_data.set_index(['trade_time','code'])
trader_data = filter_trading_time(data=trader_data, trading_sessions=trading_sessions).sort_values(by=['trade_time','code'])
```

```python
trader_data.head()
```

```python
#### 对比公式
def rolling_compare(research_data, trader_data, name, window):
    common = (research_data[['trade_time', 'code', name
                             ]].merge(trader_data[['trade_time', 'code', name]],
                                      on=["trade_time", "code"],
                                      suffixes=("_research", "_trader")))
    common["rolling_corr"] = (
        common["{}_research".format(name)].rolling(window).corr(
            common["{}_trader".format(name)]))
    return common
```

### vwap 价格对比

```python
vwap_common = rolling_compare(
    research_data=research_data, 
    trader_data=trader_data, 
    name = 'vwap', window=5)
vwap_common.head()
```

```python
vwap_common[vwap_common['rolling_corr'] < 0.9]
```

### 收盘 价格对比

```python
close_common = rolling_compare(
    research_data=research_data, 
    trader_data=trader_data, 
    name = 'close', window=5)
close_common.head()
```

```python
close_common[close_common['rolling_corr'] < 0.9]
```

### 成交量 对比

```python
volume_common = rolling_compare(
    research_data=research_data, 
    trader_data=trader_data, 
    name = 'volume', window=5)
volume_common.tail()
```

```python
volume_common[volume_common['rolling_corr'] < 0.9]
```

### 持仓量 对比

```python
openint_common = rolling_compare(
    research_data=research_data, 
    trader_data=trader_data, 
    name = 'openint', window=5)
openint_common.tail()
```

```python
openint_common[openint_common['rolling_corr'] < 0.9]
```
