```python
import pandas as pd
import numpy as np

def compare_frames(df_old, df_new, key_cols=None, float_tol=1e-9):
    left = df_old.copy()
    right = df_new.copy()

    if key_cols is None:
        key_cols = ["datetime", "vt_symbol"]

    left = left.sort_values(key_cols).reset_index(drop=True)
    right = right.sort_values(key_cols).reset_index(drop=True)

    result = {}

    # 1. 形状和列是否一致
    result["shape_equal"] = left.shape == right.shape
    result["columns_equal"] = list(left.columns) == list(right.columns)

    common_cols = [c for c in left.columns if c in right.columns]

    # 2. 每列是否一致
    column_summary = []
    diff_rows = []

    for col in common_cols:
        s1 = left[col]
        s2 = right[col]

        if pd.api.types.is_numeric_dtype(s1) and pd.api.types.is_numeric_dtype(s2):
            equal_mask = np.isclose(s1.astype(float), s2.astype(float), atol=float_tol, rtol=0, equal_nan=True)
        else:
            equal_mask = s1.fillna("__NA__").eq(s2.fillna("__NA__"))

        mismatch_count = int((~equal_mask).sum())

        column_summary.append({
            "column": col,
            "equal": mismatch_count == 0,
            "mismatch_count": mismatch_count,
        })

        if mismatch_count > 0:
            idx = np.where(~equal_mask)[0]
            for i in idx[:20]:   # 每列最多展示前20条
                row = {"row": int(i), "column": col, "old": s1.iloc[i], "new": s2.iloc[i]}
                for k in key_cols:
                    row[k] = left.iloc[i][k]
                diff_rows.append(row)

    result["column_summary"] = pd.DataFrame(column_summary)
    result["diff_rows"] = pd.DataFrame(diff_rows)

    return result
```
